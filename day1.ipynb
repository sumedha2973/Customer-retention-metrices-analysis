```python
# Function to take inputs from the user
def get_inputs():

    while True:
        try:
            # Ask for units sold
            units = input("Enter units sold (or type 'end' to stop): ")

            # Check if the user wants to stop
            if units.lower() == "end":
                return None

            # Convert units into integer
            units = int(units)

            # Ask for price per unit
            price = input("Enter price per unit (or type 'end' to stop): ")

            if price.lower() == "end":
                return None

            # Convert price into decimal number
            price = float(price)

            # Ask for sales target
            target = input("Enter sales target (or type 'end' to stop): ")

            if target.lower() == "end":
                return None

            # Convert target into decimal number
            target = float(target)

            # Return all three values
            return units, price, target

        except ValueError:
            # Handle invalid numeric input
            print("Invalid input! Please enter valid numbers.")
            print("Please enter all values again.\n")


# Function to calculate and display sales performance
def sales_performance(units, price, target):

    try:
        # Calculate total revenue
        revenue = units * price

        # Calculate target achievement
        achievement = (revenue / target) * 100

        # Display the results
        print("\n--- Sales Performance ---")
        print(f"Total Revenue: ₹{revenue:.2f}")
        print(f"Target Achievement: {achievement:.2f}%")

    except ZeroDivisionError:
        # Handle zero sales target
        print("Error! Sales target cannot be zero.")
        return False

    return True


# Main program using a while loop
while True:

    # Get inputs from the user
    data = get_inputs()

    # Stop the program if the user types 'end'
    if data is None:
        print("Program stopped. Thank you!")
        break

    # Unpack the input values
    units, price, target = data

    # Calculate sales performance
    result = sales_performance(units, price, target)

    # If target is zero, ask for inputs again
    if result == False:
        print("Please enter the details again.\n")
        continue
```

```python
# Class for Sales Performance Analysis
class SalesPerformance:

    # Method to take inputs from the user
    def get_inputs(self):

        while True:
            try:
                # Ask for units sold
                units = input("Enter units sold (or type 'end' to stop): ")

                if units.lower() == "end":
                    return None

                units = int(units)

                # Ask for price per unit
                price = input("Enter price per unit (or type 'end' to stop): ")

                if price.lower() == "end":
                    return None

                price = float(price)

                # Ask for sales target
                target = input("Enter sales target (or type 'end' to stop): ")

                if target.lower() == "end":
                    return None

                target = float(target)

                # Return input values
                return units, price, target

            except ValueError:
                print("Invalid input! Please enter valid numbers.")
                print("Try again.\n")

    # Method to calculate sales performance
    def sales_performance(self, units, price, target):

        try:
            # Calculate revenue
            revenue = units * price

            # Calculate target achievement
            achievement = (revenue / target) * 100

            # Display results
            print("\n--- Sales Performance ---")
            print(f"Total Revenue: ₹{revenue:.2f}")
            print(f"Target Achievement: {achievement:.2f}%")

            return True

        except ZeroDivisionError:
            print("Error! Sales target cannot be zero.")
            return False


# Create an object of the class
sales = SalesPerformance()

# Main loop
while True:

    # Call get_inputs() using the object
    data = sales.get_inputs()

    # Stop if user enters 'end'
    if data is None:
        print("Program stopped. Thank you!")
        break

    # Unpack input values
    units, price, target = data

    # Call sales_performance() using the object
    result = sales.sales_performance(units, price, target)

    # Ask for inputs again if target is zero
    if result == False:
        print("Please enter the details again.\n")
        continue
```
